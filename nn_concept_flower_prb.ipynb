```python
import numpy as np
import matplotlib.pyplot as plt

# Input data (features)
inputs = np.array([
    [3, 1.5],
    [2, 1],
    [4, 1.5],
    [3, 1],
    [3.5, 0.5],
    [2, 0.5],
    [5.5, 1],
    [1,1]
])

# Target data (labels)
targets = np.array([[1, 0, 1, 0, 1, 0, 1, 0]]).T  # Shape: 8x1

# Mystery flower to classify
mystery_flower = [1.5, 1]

# Initialize weights and bias
np.random.seed(1)
weights = np.random.randn(2, 1)  # Shape: 2x1
bias = np.random.randn()         # Scalar

# Print initial weights and bias
print("Initial weights:")
print(weights)
print("Initial bias:")
print(bias)

# Activation functions
def sigmoid(x):
    return 1 / (1 + np.exp(-x))

def sigmoid_derivative(x):
    return x * (1 - x)

# Forward pass
def think(inputs, weights, bias):
    return sigmoid(np.dot(inputs, weights) + bias)

# Training function
def train(inputs, targets, weights, bias, learning_rate=0.1, epochs=1):
    for epoch in range(epochs):
        #print(f"\nEpoch {epoch + 1}:")
        for i in range(len(inputs)):
            # Select a single training example
            input_example = inputs[i].reshape(1, -1)
            true_output = targets[i]

            # Forward pass
            output = think(input_example, weights, bias)

            # Calculate error
            error = true_output - output



            # Calculate adjustments
            adjustments = error * sigmoid_derivative(output)

            # Update weights and bias
            weights += np.dot(input_example.T, adjustments) * learning_rate
            bias += np.sum(adjustments) * learning_rate
            if(epoch % 1000 == 0):
              # Print updated weights and bias for this example
              print(f"  After training example {i + 1}:")
              print(f"    Weights:\n{weights}")
              print(f"    Bias:\n{bias}")

    return weights, bias

# Train the model
weights, bias = train(inputs, targets, weights, bias, learning_rate=0.1, epochs=10000)

# Classify the mystery flower
output = think(np.array(mystery_flower).reshape(1, -1), weights, bias)
print("\nPrediction for the mystery flower:")
print(f"Output: {output}")
print("Class: 1" if output >= 0.5 else "Class: 0")


```
